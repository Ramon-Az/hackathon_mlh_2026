# Primeira chamada de API

```python
# pip install backboard-sdk python-dotenv
import asyncio
import os
from pathlib import Path
from typing import cast

from dotenv import load_dotenv
from backboard import BackboardClient
from backboard.models import ChatMessagesResponse

path_dir = Path.cwd()
if not (path_dir / ".env").exists() and path_dir.parent != path_dir:
    path_dir = path_dir.parent

load_dotenv(dotenv_path=path_dir / ".env")

async def main():
    api_key = os.getenv("API_KEY")
    if not api_key:
        raise ValueError(f"API_KEY não encontrado em {path_dir / '.env'}")

    client = BackboardClient(api_key=api_key)

    # Send a message — thread and assistant are auto-created
    response = cast(
        ChatMessagesResponse,
        await client.send_message(
            "Hello! I'm excited to get started.",
            memory="Auto",
            stream=False,
        ),
    )
    print(response.content)

    # Continue the conversation using the returned thread_id
    response = cast(
        ChatMessagesResponse,
        await client.send_message(
            "What can you help me with?",
            thread_id=response.thread_id,
            stream=False,
        ),
    )
    print(response.content)

await main()
```
